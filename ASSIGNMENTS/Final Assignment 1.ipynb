```python
#question no 4
def addition(num1,num2):
    return num1+num2
def substraction(num1,num2):
    return num1-num2    
def multiplication(num1,num2):
    return num1*num2
def division(num1,num2):
    return num1/num2  
def modulus(num1,num2):
    return num1%num2    

operations={
    "Addition:  ":addition,
    "SUBStraction:  ":substraction,
    "multiplication:  ":multiplication,
    "division:  ":division,
    "Modulus:  ":modulus}
operation=input("ENTER an operation you want to perform [+,-,*,/,%] : ")
flag=True
while flag:
    if(operation=="+"):
        a=int(input("Enter a number: "))
        b=int(input("Enter 2nd number: "))
        result=addition(a,b)
        print(result)
    elif(operation=="-"):
        a=int(input("Enter a number: "))
        b=int(input("Enter 2nd number: "))
        result=substraction(a,b)
        print(result)
    elif(operation=="*"):
        a=int(input("Enter a number: "))
        b=int(input("Enter 2nd number: "))   
        result=multiplication(a,b)
        print(result)
    elif(operation=="/"):  
        a=int(input("Enter a number: "))
        b=int(input("Enter 2nd number: "))   
        result=division(a,b)
        print(result)
    else:
        a=int(input("Enter a number: "))
        b=int(input("Enter 2nd number: "))   
        result=modulus(a,b)
        print(result)
    console="do you want to continue (y/n): "
    print(input(console))
    if console!="y":
        print("okii bye!")
        flag=False
```

```python
# python assignment Question no.1
# Student result management system 
# part 1 and 2

def result(marks1, marks2, marks3):
    total_marks = marks1 + marks2 + marks3
    percentage = total_marks / 3
    if percentage >= 80:
        grade = "GRADE A!congratulations bro!"
    elif percentage >= 70:
        grade = "GRADE B"
    elif percentage >= 60:
        grade = "GRADE C"
    elif percentage >= 50:  
        grade = "GRADE D"
    else:
        grade = "GRADE F!better luck next time"
    return grade, total_marks, percentage

students_list = []

for i in range(6):
    print("Enter details for student ")
    student_name = input("Enter student's name : ")
    roll_no = int(input("Enter student's roll no: "))
    python_marks = int(input("Enter python Marks: "))
    Artificial_intelligence = int(input("Enter AI marks: "))
    Maths = int(input("Enter Maths marks: "))
    
    grd, tot_marks, perc = result(python_marks, Artificial_intelligence, Maths)
    student_info = {
        "NAME": student_name,
        "ROLL NO": roll_no,
        "PYTHON": python_marks,
        "Artificial Intelligence": Artificial_intelligence,
        "Maths": Maths,
        "total_marks": tot_marks,
        "percentage": perc,
        "grade": grd
    }
    
    students_list.append(student_info)
# Display results
for student_info in students_list:
    print("Name:", student_info["NAME"], 
          "ROLL NO:", student_info["ROLL NO"], 
          "Total:", student_info["total_marks"], 
          "Percentage:", round(student_info["percentage"], 2), 
          "Grade:", student_info["grade"])

# Find highest percentage
highest = students_list[0] #lets assume the first one is highest
for student_info in students_list:  
    if student_info["percentage"] > highest["percentage"]:
        highest = student_info

print("the top student : ")
print(highest['NAME'])
print("percentage: ")
print(highest['percentage'])
# Save to file
file = open("result.txt", "w")
for student_info in students_list:
    file.write("Name: " + student_info["NAME"] + "\n")
    file.write("Roll No: " + str(student_info["ROLL NO"]) + "\n")
    file.write("Total: " + str(student_info["total_marks"]) + "\n")
    file.write("Percentage: " + str((student_info["percentage"])) + "\n")
    file.write("Grade: " + student_info["grade"] + "\n")
    file.write("-" * 30 + "\n")

file.close()
print("Results successfully saved to 'result.txt'.")
```

```python
#question 5
menu = {
    1: {"item": "Burger", "price": 500},
    2: {"item": "Pizza", "price": 800},
    3: {"item": "Fries", "price": 250},
    4: {"item": "Drink", "price": 150}
}
order = []
print("OUR RESTURANT MENU is given below")
for key, val in menu.items():
    print(str(key) + ". " + val['item'] + " - Rs. " + str(val['price']))
while True:
    choice = int(input("enter item number to order & 0 to finish: "))
    if choice == 0:
        break
    if choice in menu:
        quantity = int(input("Dear customer! enter quantity of your choice: "))
        item_total = menu[choice]["price"] * quantity
        order.append({
            "item": menu[choice]["item"],
            "price": menu[choice]["price"],
            "quantity": quantity,
            "itemmTotal": item_total
        })
    else:
        print("INVALID CHOICE")
final_bill = sum(item["itemmTotal"] for item in order)
print("\n================ FINAL BILL ================")
for item in order:
    print(item['item'] + " x " + str(item['quantity']) + " = Rs. " + str(item['itemmTotal']))
print("Grand Total: Rs. " + str(final_bill))
file = open("orders.txt", "w")
file.write("--- RESTAURANT BILL ---\n")
for item in order:
    file.write(item['item'] + " | Qty: " + str(item['quantity']) + " | Total: Rs. " + str(item['itemmTotal']) + "\n")
file.write("Grand Total: Rs. " + str(final_bill) + "\n")
file.close()
print("Bill successfully saved in 'orders.txt'.")
```

```python
#question no 2
def analyze_sentence(sentence):
    words = sentence.split()
    total_words = len(words)
    total_chars = len(sentence)

    longest = words[0]
    shortest = words[0]

    for w in words:
        if len(w) > len(longest):
            longest = w

        if len(w) < len(shortest):
            shortest = w

    vowels = 0

    for ch in sentence.lower():
        if ch in "aeiou":
            vowels += 1

    unique_words = set(words)
    sorted_words = sorted(unique_words)

    return total_words, total_chars, longest, shortest, vowels, unique_words, sorted_words


sentence = input("Enter a sentence: ")

tw, tc, lw, sw, v, uw, sorted_w = analyze_sentence(sentence)

print("Total words:", tw)
print("Total characters:", tc)
print("Longest word:", lw)
print("Shortest word:", sw)
print("Vowels:", v)
print("Unique words:", uw)
print("Alphabetical:", sorted_w)

with open("word_analysis.txt", "w") as f:
    f.write("Sentence: " + sentence + "\n")
    f.write("Total words: " + str(tw) + "\n")
    f.write("Total characters: " + str(tc) + "\n")
    f.write("Longest word: " + lw + "\n")
    f.write("Shortest word: " + sw + "\n")
    f.write("Vowels: " + str(v) + "\n")
    f.write("Unique words: " + str(uw) + "\n")
    f.write("Alphabetical order: " + ", ".join(sorted_w) + "\n")

print("Analysis saved in word_analysis.txt")
```
